```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
OUT = "data"

TRUE_EFFECT = 0.18   # ~ +18% exports for treated firms, after treatment

# ---------------------------------------------------------------------------
# PHASE 1: PILOT -> RCT
# ---------------------------------------------------------------------------
def generate_rct(n_firms=400):
    firm_id = np.arange(1, n_firms + 1)
    baseline_size = rng.normal(10, 1.2, n_firms)
    region = rng.integers(1, 6, n_firms)
    treated = rng.binomial(1, 0.5, n_firms)  # pure coin flip

    noise = rng.normal(0, 0.5, n_firms)
    region_effect = {r: rng.normal(0, 0.15) for r in range(1, 6)}
    region_fx = np.array([region_effect[r] for r in region])

    log_exports = baseline_size + region_fx + TRUE_EFFECT * treated + noise

    df = pd.DataFrame({
        "firm_id": firm_id, "region": region,
        "baseline_size": baseline_size, "treated": treated,
        "log_exports": log_exports,
    })
    df.to_csv(f"{OUT}/phase1_rct.csv", index=False)
    return df


# ---------------------------------------------------------------------------
# PHASE 2: REGIONAL ROLLOUT -> DiD
# ---------------------------------------------------------------------------
def generate_did(n_firms=600, n_regions=8, n_periods=12, policy_start=7):
    firm_id = np.repeat(np.arange(1, n_firms + 1), n_periods)
    period = np.tile(np.arange(1, n_periods + 1), n_firms)
    region = np.repeat(rng.integers(1, n_regions + 1, n_firms), n_periods)

    treated_regions = set(range(1, n_regions // 2 + 1))
    treated_region_flag = np.isin(region, list(treated_regions)).astype(int)
    post = (period >= policy_start).astype(int)
    treated_post = treated_region_flag * post

    firm_fx = rng.normal(0, 0.8, n_firms)
    firm_fx_full = np.repeat(firm_fx, n_periods)

    time_trend = 0.03 * period
    region_trend = {r: rng.normal(0.0, 0.02) for r in range(1, n_regions + 1)}
    region_trend_full = np.array([region_trend[r] for r in region]) * period

    noise = rng.normal(0, 0.4, len(firm_id))

    log_exports = (10 + firm_fx_full + time_trend + region_trend_full
                   + TRUE_EFFECT * treated_post + noise)

    df = pd.DataFrame({
        "firm_id": firm_id, "period": period, "region": region,
        "treated_region": treated_region_flag, "post": post,
        "treated_post": treated_post, "log_exports": log_exports,
    })
    df.to_csv(f"{OUT}/phase2_did.csv", index=False)
    return df


# ---------------------------------------------------------------------------
# PHASE 3: NATIONAL ROLLOUT -> RDD
# ---------------------------------------------------------------------------
def generate_rdd(n_firms=1000, cutoff=50):
    firm_id = np.arange(1, n_firms + 1)
    running_var = rng.uniform(0, 100, n_firms)
    treated = (running_var >= cutoff).astype(int)

    smooth_part = 8 + 0.02 * running_var + 0.0003 * (running_var - 50) ** 2
    noise = rng.normal(0, 0.35, n_firms)
    log_exports = smooth_part + TRUE_EFFECT * treated + noise

    df = pd.DataFrame({
        "firm_id": firm_id, "running_var": running_var,
        "cutoff": cutoff, "treated": treated, "log_exports": log_exports,
    })
    df.to_csv(f"{OUT}/phase3_rdd.csv", index=False)
    return df


rct = generate_rct()
did = generate_did()
rdd = generate_rdd()
print("Generated:")
print(f"  phase1_rct.csv  -> {rct.shape}")
print(f"  phase2_did.csv  -> {did.shape}")
print(f"  phase3_rdd.csv  -> {rdd.shape}")
print(f"\nTrue treatment effect used in all 3 phases: {TRUE_EFFECT}")
```

```python
import pandas as pd
import statsmodels.formula.api as smf

df = pd.read_csv("data/phase1_rct.csv")

balance = smf.ols("baseline_size ~ treated", data=df).fit()
model = smf.ols("log_exports ~ treated", data=df).fit(cov_type="HC1")

print("=== Balance check (baseline_size ~ treated) ===")
print(balance.params, "\np-value on 'treated':", balance.pvalues["treated"])

print("\n=== RCT estimate (log_exports ~ treated) ===")
print(model.summary())

with open("outputs/results_summary.txt", "w") as f:
    f.write("TRUE EFFECT USED IN SIMULATION: 0.18 (log points, i.e. ~19.7% increase in exports)\n\n")
    f.write("PHASE 1 - RCT\n")
    f.write(f"Estimated effect: {model.params['treated']:.4f}  "
            f"(SE {model.bse['treated']:.4f}, 95% CI "
            f"[{model.conf_int().loc['treated', 0]:.4f}, "
            f"{model.conf_int().loc['treated', 1]:.4f}])\n")
    f.write(f"Balance check p-value on baseline_size: {balance.pvalues['treated']:.4f}\n\n")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

df = pd.read_csv("data/phase2_did.csv")

trend = df.groupby(["period", "treated_region"])["log_exports"].mean().reset_index()

plt.figure(figsize=(7, 4.5))
for grp, label, style in [(1, "Treated regions", "-o"), (0, "Control regions", "-s")]:
    sub = trend[trend.treated_region == grp]
    plt.plot(sub.period, sub.log_exports, style, label=label)
plt.axvline(6.5, color="gray", linestyle="--", label="Policy start")
plt.xlabel("Period")
plt.ylabel("Mean log(exports)")
plt.title("Phase 2: Parallel trends check")
plt.legend()
plt.tight_layout()
plt.savefig("outputs/did_parallel_trends.png", dpi=140)
plt.show()

model = smf.ols(
    "log_exports ~ treated_post + C(firm_id) + C(period)", data=df
).fit(cov_type="cluster", cov_kwds={"groups": df["region"]})

effect = model.params["treated_post"]
se = model.bse["treated_post"]
ci = model.conf_int().loc["treated_post"]

print(f"DiD estimate (treated_post): {effect:.4f} (SE {se:.4f})")
print(f"95% CI: [{ci[0]:.4f}, {ci[1]:.4f}]")

with open("outputs/results_summary.txt", "a") as f:
    f.write("PHASE 2 - DiD (TWFE, region-clustered SE)\n")
    f.write(f"Estimated effect: {effect:.4f}  (SE {se:.4f}, 95% CI [{ci[0]:.4f}, {ci[1]:.4f}])\n\n")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

df = pd.read_csv("data/phase3_rdd.csv")
cutoff = df["cutoff"].iloc[0]
df["centered"] = df["running_var"] - cutoff

bw = 15
sub = df[np.abs(df["centered"]) <= bw].copy()
sub["treated_x_centered"] = sub["treated"] * sub["centered"]

model = smf.wls(
    "log_exports ~ centered + treated + treated_x_centered",
    data=sub,
    weights=(1 - np.abs(sub["centered"]) / bw)
).fit(cov_type="HC1")

effect = model.params["treated"]
se = model.bse["treated"]
ci = model.conf_int().loc["treated"]

print(f"RDD estimate (jump at cutoff): {effect:.4f} (SE {se:.4f})")
print(f"95% CI: [{ci[0]:.4f}, {ci[1]:.4f}]  |  bandwidth = +/-{bw}")

df["bin"] = (df["running_var"] // 4) * 4 + 2
binned = df.groupby("bin")["log_exports"].mean().reset_index()

plt.figure(figsize=(7, 4.5))
plt.scatter(binned.bin, binned.log_exports, color="steelblue", s=25, label="Binned means")

for side, color in [(sub[sub.treated == 0], "darkorange"), (sub[sub.treated == 1], "green")]:
    xs = np.linspace(side.centered.min(), side.centered.max(), 50) + cutoff
    coefs = np.polyfit(side.centered, side.log_exports, 1)
    plt.plot(xs, np.polyval(coefs, xs - cutoff), color=color, linewidth=2)

plt.axvline(cutoff, color="gray", linestyle="--", label="Cutoff")
plt.xlabel("Readiness score (running variable)")
plt.ylabel("log(exports)")
plt.title("Phase 3: RDD - discontinuity at the cutoff")
plt.legend()
plt.tight_layout()
plt.savefig("outputs/rdd_discontinuity.png", dpi=140)
plt.show()

with open("outputs/results_summary.txt", "a") as f:
    f.write("PHASE 3 - RDD (local linear, triangular kernel, bandwidth=15)\n")
    f.write(f"Estimated effect: {effect:.4f}  (SE {se:.4f}, 95% CI [{ci[0]:.4f}, {ci[1]:.4f}])\n\n")
```
